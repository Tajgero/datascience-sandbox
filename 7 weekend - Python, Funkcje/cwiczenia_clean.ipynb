### Ćwiczenie 1 – Netto i brutto dla produktu

Napisz funkcję, która dla podanej ceny netto i stawki VAT obliczy cenę brutto.

- Wejście: `net_amount` (float lub int), `vat_rate` (float, np. 0.23).

```python
# Ćwiczenie 1
def calculate_gross(net_amount, vat_rate):
    """Zwróć kwotę brutto dla podanej kwoty netto i stawki VAT.

    net_amount: kwota netto (liczba)
    vat_rate: stawka VAT, np. 0.23
    """
    gross_amount = net_amount * (1 + vat_rate)
    return gross_amount

print(calculate_gross(100, 0.23))
```

### Ćwiczenie 2 – Średnia wartość koszyka klienta

Napisz funkcję, która policzy średnią wartość zamówienia dla wybranego klienta na podstawie `orders_data`.

- Wejście: `orders_data` oraz `customer_id`.
- Wyjście: średnia wartość `total_net` dla danego klienta (float) lub `0` jeśli brak zamówień.

Na razie możesz założyć, że `total_net` jest wypełnione.
Jeśli nie jest – tymczasowo policz sumę wartości pozycji w `items`.


```python
# Ćwiczenie 2
def average_order_value(orders_data, customer_id):
    """Zwróć średnią wartość zamówienia dla danego klienta.

    Jeśli klient nie ma zamówień, zwróć 0.
    Jeśli total_net jest puste (None), liczy sumę pozycji z items
    (quantity * price) jako wartość zamówienia.
    """
    total_sum = 0
    order_count = 0

    for order in orders_data.values():
        if order["customer_id"] != customer_id:
            continue

        if order["total_net"] is not None:
            order_value = order["total_net"]
        else:
            order_value = 0
            for item in order["items"]:
                order_value += item["quantity"] * item["price"]

        total_sum += order_value
        order_count += 1

    if order_count == 0:
        return 0

    return total_sum / order_count

print(average_order_value(orders_data, "C001"))
print(average_order_value(orders_data, "C002"))
print(average_order_value(orders_data, "C003"))
```

Dodaj prosty docstring opisujący działanie funkcji.


```python
# Ćwiczenie 3
def filter_active_customers(customers_data: dict, country: str ="PL") -> list:
    """Zwróć listę ID aktywnych klientów z danego kraju.
    
    Args:
        customers_data (dict): Słownik z danymi klientów.
        country (str): Kraj klientów (domyślnie "PL").

    Returns:
        list: Lista ID aktywnych klientów z danego kraju.
    """
    active_customers = []
    for id, customer in customers_data.items():
        if (customer["country"] == country) and customer["is_active"]:
            active_customers.append(id)
    return active_customers

print(filter_active_customers(customers_data, country="PL"))
```

### Ćwiczenie 6 – Zmienna liczba rabatów (*args)

Napisz funkcję, która przyjmie cenę bazową oraz dowolną liczbę rabatów procentowych (np. 0.1, 0.05)
i zwróci cenę po zastosowaniu wszystkich rabatów.

- Wejście: `base_price`, `*discounts`.
- Wyjście: cena końcowa.

Pamiętaj, że `discounts` wewnątrz funkcji to krotka.


```python
# Ćwiczenie 6
def apply_discounts(base_price, *discounts):
    """Zastosuj dowolną liczbę rabatów procentowych do ceny bazowej."""
    total_discount = sum(discounts)
    return base_price - (base_price * total_discount)

# Przykłady:
print(apply_discounts(100))
print(apply_discounts(100, 0.1))
print(apply_discounts(100, 0.1, 0.05))
```

```python
# Ćwiczenie 12
def create_discount_func(discount):
    """Zwróć funkcję, która przyjmie price i zwróci cenę po rabacie."""

    def apply_discount(price):
        return price * (1 - discount)

    return apply_discount


discount_10 = create_discount_func(0.1)
print(discount_10(100))  # 90.0
discount_20 = create_discount_func(0.2)
print(discount_20(100))  # 80.0
```

### Ćwiczenie 15 – Funkcje jako dane – prosty pipeline

Zbuduj prosty "pipeline" funkcji czyszczących nazwy produktów.

Przykładowe kroki:
- usuwanie spacji z początku i końca,
- zamiana na małe litery,
- zamiana wielokrotnych spacji na pojedynczą.

Umieść te kroki w liście funkcji i zastosuj je po kolei do każdej nazwy.


```python
# Ćwiczenie 15
def strip_spaces(name):
    return name.strip()

def to_lower(name):
    return name.lower()
    
def collapse_spaces(name):
    return name.replace("  ", " ")

pipeline = [strip_spaces, to_lower, collapse_spaces]

raw_product_names = ["  ala  ma kot a", " ", "  ", "ale jAJA"]

cleaned = []
for name in raw_product_names:
    temp = name
    for function in pipeline:
        temp = function(temp)
    cleaned.append(temp)

print(cleaned)
```

- Wejście: `orders_data`.

```python
# Ćwiczenie 20
def count_orders_per_customer(orders_data):
    """Zwróć słownik {customer_id: liczba_zamówień}."""
    out = {}
    for order in orders_data.values():
        customer = order["customer_id"]
        
        if customer in out:
            out[customer] += 1
        else:
            out[customer] = 1
            
    return out

# Przykład:
print(count_orders_per_customer(orders_data))
```
