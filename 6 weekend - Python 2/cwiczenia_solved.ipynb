Pracujemy z truthiness, porownaniami i przynaleznoscia do kolekcji.

### Zadanie 5 - Bool i truthiness
- Sprawdz dzialanie `bool()` dla roznych wartosci `discount_code`.
- Napisz warunek wypisujacy komunikat, gdy kodu brak.

### Zadanie 6 - Porownania wartosci zamowien
- Na podstawie `order_value` klasyfikuj zamowienia jako `small/medium/large`.

### Zadanie 7 - Porownanie z None
- Dla kazdego zamowienia sprawdz, czy posiada kod rabatowy.

### Zadanie 8 - Operator `in` w kolekcjach

### Zadanie 9 - Porownania leksykalne stringow
- Posortuj statusy i sprawdz porownania miedzy nimi.


```python
# ZADANIE 5 - Bool i truthiness

# 1) pusty string
discount_code = ''
print('discount_code = "" -> bool:', bool(discount_code))

# 2) niepusty string
discount_code = 'BLACK20'
print("discount_code = 'BLACK20' -> bool:", bool(discount_code))

# 3) None
discount_code = None
print('discount_code = None -> bool:', bool(discount_code))

# prosty warunek dla braku kodu
if not discount_code:
    print('Brak kodu rabatowego')
else:
    print('Mamy kod rabatowy:', discount_code)
```

```python
# ZADANIE 6 - Porownania wartosci zamowien

order_value = 450.0

if order_value < 200:
    size = 'small'
elif order_value <= 500:
    size = 'medium'
else:
    size = 'large'

print('Wartosc zamowienia:', order_value, '-> kategoria:', size)
```

```python
# ZADANIE 7 - Porownanie z None

for order in orders:
    code = order['discount_code']
    if code is None:
        print('Order', order['order_id'], '- brak kodu rabatowego')
    else:
        print('Order', order['order_id'], '- kod rabatowy:', code)
```

```python
# ZADANIE 8 - Operator in w kolekcjach

for order in orders:
    country = order['country']
    if country in blocked_countries:
        print('Order', order['order_id'], '- kraj', country, 'jest na liscie zablokowanych')
    else:
        print('Order', order['order_id'], '- kraj', country, 'jest OK')
```

```python
# ZADANIE 9 - Porownania leksykalne stringow

statuses = ['completed', 'cancelled', 'returned', 'pending']
print('Statusy oryginalne:', statuses)

sorted_statuses = sorted(statuses)
print('Statusy posortowane:', sorted_statuses)

# przyklady porownan
status_a = 'cancelled'
status_b = 'completed'
print('"cancelled" < "completed"?', status_a < status_b)

status_a = 'pending'
status_b = 'returned'
print('"pending" < "returned"?', status_a < status_b)
```

## Cwiczenia 10-12: logika, `and` / `or` / `not`, operatory bitowe

### Zadanie 10 - Logika `and` / `or` w filtrach
- Filtrujemy zamowienia wedlug kraju, wartosci i dodatkowych warunkow.

### Zadanie 11 - `is` vs `==`
- Porownujemy listy, zeby zobaczyc roznice miedzy rownoscia a tozsamoscia.

### Zadanie 12 (opcjonalne) - Operatory bitowe i flagi
- Prosta zabawa z bitami dla flag statusu.


```python
# ZADANIE 10 - Logika and / or w filtrach

high_value_pl_orders = []
high_value_other_orders = []

for order in orders:
    value = order['order_value']
    country = order['country']

    # duze zamowienie z Polski (> 500)
    if value > 500 and country == 'PL':
        high_value_pl_orders.append(order)

    # duze zamowienie z innego kraju (wartosc > 1000)
    if value > 1000 and country != 'PL':
        high_value_other_orders.append(order)

print('Duze zamowienia z Polski (>500):')
for o in high_value_pl_orders:
    print(o)

print('\nDuze zamowienia z innych krajow (>1000):')
for o in high_value_other_orders:
    print(o)

# przyklad z uzyciem and/or w jednym warunku
print('\nZamowienia spelniajace zlozony warunek:')
for order in orders:
    value = order['order_value']
    country = order['country']
    if (value > 500 and country == 'PL') or (value > 1000 and country != 'PL'):
        print('Wybrane zamowienie:', order)
```

```python
# ZADANIE 11 - is vs ==

a = [1, 2, 3]
b = [1, 2, 3]

print('a == b?', a == b)   # wartosci sa takie same
print('a is b?', a is b)   # dwa rozne obiekty w pamieci

# teraz b wskazuje na ten sam obiekt co a
b = a
print('\nPo przypisaniu b = a:')
print('a == b?', a == b)   # nadal te same wartosci
print('a is b?', a is b)   # ten sam obiekt

# Wniosek (w komentarzu):
# - '==' porownuje wartosci elementow,
# - 'is' porownuje tozsamosc (adres obiektu w pamieci).
```

```python
# ZADANIE 12 (opcjonalne) - Operatory bitowe

# Przykladowe flagi (bit 0, 1, 2):
# bit 0 - completed
# bit 1 - cancelled
# bit 2 - returned

completed_flag = 0b001
cancelled_flag = 0b010
returned_flag = 0b100

# przykladowy status: zamowienie zrealizowane i zwrocone (teoretycznie) -> bity 0 i 2
some_status = 0b101

print('some_status w binarnym:', bin(some_status))
print('Czy completed?', bool(some_status & completed_flag))
print('Czy cancelled?', bool(some_status & cancelled_flag))
print('Czy returned?', bool(some_status & returned_flag))
```

## Cwiczenia 13-16: instrukcje warunkowe, inline if, match-case

### Zadanie 13 - Warunki biznesowe: tagowanie klienta
- Tag klienta na podstawie liczby zamowien i lacznej wydanej kwoty.

### Zadanie 14 - Inline if-else
- Ustawienie typu dostawy w zaleznosci od wartosci zamowienia.

### Zadanie 15 - Match-case: mapowanie statusow
- Wysylamy komunikat klientowi na podstawie statusu zamowienia.

### Zadanie 16 (opcjonalne) - Match-case: kategorie produktow
- Przypisanie stawki podatku do kategorii produktu.


```python
# ZADANIE 13 - Warunki biznesowe: tagowanie klienta

orders_count = 4
total_spent = 6200.0

customer_tag = 'STANDARD'

# najpierw sprawdzamy VIP (wysoka kwota), potem LOYAL, potem NEW
if total_spent > 5000:
    customer_tag = 'VIP'
elif orders_count > 3:
    customer_tag = 'LOYAL'
elif orders_count == 1:
    customer_tag = 'NEW'

print('Liczba zamowien:', orders_count)
print('Lacznie wydane:', total_spent)
print('Tag klienta:', customer_tag)
```

```python
# ZADANIE 14 - Inline if-else

order_value = 450.0

# klasyczny if / else
if order_value > 300:
    shipping_type = 'free'
else:
    shipping_type = 'paid'

# wersja inline (operator trojargumentowy)
shipping_type_inline = 'free' if order_value > 300 else 'paid'

print('shipping_type:', shipping_type)
print('shipping_type_inline:', shipping_type_inline)
```

```python
# ZADANIE 15 - Match-case: mapowanie statusow

status = 'completed'

match status:
    case 'completed':
        customer_message = 'Zamowienie zrealizowane'
    case 'cancelled':
        customer_message = 'Zamowienie anulowane'
    case 'returned':
        customer_message = 'Zamowienie zwrocone'
    case 'pending':
        customer_message = 'Zamowienie w trakcie realizacji'
    case _:
        customer_message = 'Status nieznany'

print('Status:', status)
print('Komunikat do klienta:', customer_message)
```

```python
# ZADANIE 16 (opcjonalne) - Match-case: kategorie produktow

category = 'electronics'

match category:
    case 'electronics':
        tax_rate = 23
    case 'fashion':
        tax_rate = 23
    case 'home':
        tax_rate = 8
    case _:
        tax_rate = 23

print('Kategoria produktu:', category)
print('Stawka podatku (%):', tax_rate)
```

```python
# ZADANIE 19 - Iteracja po stringu

discount_code = 'BLACKFRIDAY20'
digits_count = 0

for char in discount_code:
    if char.isdigit():
        digits_count += 1

print('Kod rabatowy:', discount_code)
print('Liczba cyfr w kodzie:', digits_count)
```

## Cwiczenia 21-24: petla `while`, `zip`, `input`, list comprehension

### Zadanie 21 - Petla `while` - licznik prob logowania
- Maksymalnie 3 proby czegos (np. logowania / wprowadzenia kodu).

### Zadanie 22 - Funkcja `input()` - prosty interfejs tekstowy
- Filtr zamowien na podstawie progu wprowadzonego przez uzytkownika.

### Zadanie 23 - `zip` - laczenie list
- Wypisanie par `customer_id` - `score`.

### Zadanie 24 - List comprehension - filtry zamowien
- Filtrowanie zamowien klasycznie i jako list comprehension.


```python
# ZADANIE 21 - Petla while - licznik prob

attempts = 0
max_attempts = 3

while attempts < max_attempts:
    print('Proba nr', attempts + 1)
    # tutaj moglaby byc logika np. prosba o haslo / kod
    attempts += 1

print('Koniec prob (wykorzystano wszystkie mozliwe)')
```

```python
# ZADANIE 22 - Funkcja input() - prosty interfejs tekstowy
# Uwaga: uruchomienie tej komorki bedzie czekac na dane od uzytkownika.

threshold_str = input('Podaj minimalna wartosc zamowienia: ')

try:
    threshold = float(threshold_str)
except ValueError:
    print('Niepoprawna liczba, ustawiam domyslny prog 0')
    threshold = 0.0

print('Zamowienia o wartosci wiekszej niz', threshold, ':')
for order in orders:
    if order['order_value'] > threshold:
        print(order)
```

```python
# ZADANIE 23 - zip - laczenie list

customer_ids = [1001, 1002, 1003]
customer_scores = [0.8, 0.3, 0.5]

for customer_id, score in zip(customer_ids, customer_scores):
    print('Klient', customer_id, 'ma score', score)
```

```python
# ZADANIE 24 - List comprehension - filtry zamowien

# klasyczna wersja z petla
big_orders = []
for order in orders:
    if order['order_value'] > 500:
        big_orders.append(order)

print('Duze zamowienia (petla):')
for o in big_orders:
    print(o)

# wersja list comprehension
big_orders_comp = [order for order in orders if order['order_value'] > 500]

print('\nDuze zamowienia (list comprehension):')
for o in big_orders_comp:
    print(o)
```

## Cwiczenia 25-26: mini-projekty

Laczymy wiekszosc poznanych elementow w bardziej spojne zadania.

### Zadanie 25 - Mini-projekt 1: raport dzienny zamowien
- Obliczamy podstawowe metryki na podstawie listy `orders`.

### Zadanie 26 - Mini-projekt 2: tagowanie ryzyka transakcji
- Okreslanie poziomu ryzyka zamowien na podstawie prostych reguł biznesowych.


```python
# ZADANIE 25 - Mini-projekt 1: raport dzienny zamowien

total_orders = 0
total_revenue = 0.0
completed_count = 0
returned_or_cancelled_count = 0
pl_orders_count = 0
big_orders = []  # np. powyzej 500

for order in orders:
    total_orders += 1
    total_revenue += order['order_value']

    if order['status'] == 'completed':
        completed_count += 1
    if order['status'] in ['cancelled', 'returned']:
        returned_or_cancelled_count += 1

    if order['country'] == 'PL':
        pl_orders_count += 1

    if order['order_value'] > 500:
        big_orders.append(order)

print('RAPORT DZIENNY ZAMOWIEN')
print('------------------------')
print('Liczba zamowien:', total_orders)
print('Laczna wartosc zamowien:', total_revenue)
print('Zamowienia zrealizowane:', completed_count)
print('Zamowienia anulowane/zwrocone:', returned_or_cancelled_count)
print('Zamowienia z Polski:', pl_orders_count)
print('Duze zamowienia (> 500):')
for o in big_orders:
    print(' -', o)
```

```python
# ZADANIE 26 - Mini-projekt 2: tagowanie ryzyka transakcji

risked_orders = []  # lista krotek (order_id, risk_level)

for order in orders:
    # startowo LOW = 0, MEDIUM = 1, HIGH = 2
    risk_score = 0

    # jesli kraj na liscie blocked_countries -> co najmniej MEDIUM
    if order['country'] in blocked_countries:
        risk_score = max(risk_score, 1)

    # jesli duza wartosc i pierwsze zamowienie -> HIGH
    if order['order_value'] > 1000 and order['is_first_order']:
        risk_score = 2

    # jesli metoda platnosci jest ryzykowna -> podnosimy poziom o 1 (do max 2)
    if order['payment_type'] in high_risk_payment_types:
        risk_score = min(risk_score + 1, 2)

    # mapowanie score -> etykieta tekstowa
    if risk_score == 0:
        risk_level = 'LOW'
    elif risk_score == 1:
        risk_level = 'MEDIUM'
    else:
        risk_level = 'HIGH'

    risked_orders.append((order['order_id'], risk_level))

print('Poziomy ryzyka dla zamowien:')
for order_id, risk_level in risked_orders:
    print('Order', order_id, '->', risk_level)
```
