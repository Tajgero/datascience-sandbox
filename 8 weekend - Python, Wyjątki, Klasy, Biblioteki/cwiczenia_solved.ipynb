````python
# Sekcja 1 – Wyjatki i obsluga bledow

# Zadanie 1.1 – rozpoznawanie typow bledow
# Odkomentuj kolejne fragmenty i zaobserwuj, jaki blad powstaje

# if True
#     print('hello')  # SyntaxError – blad skladni

# my_list = [1, 2, 3]
# print(my_list[10])  # IndexError

# product = {'name': 'Laptop'}
# print(product['price'])  # KeyError

# Zadanie 1.2 – bezpieczne pobieranie amount

def print_order_amounts(orders_list):
    """Wypisuje kwoty zamowien, obslugujac brak pola amount."""
    for order in orders_list:
        try:
            print(f"Kwota zamowienia {order.get('order_id')}: {order['amount']}")
        except KeyError:
            print(f"Brakuje pola 'amount' w zamowieniu {order.get('order_id')}")


# Zadanie 1.3 – parsowanie liczby dni

def parse_days(text):
    """Zwraca liczbe dni jako int lub None w przypadku bledu."""
    try:
        days = int(text)
        if days < 0:
            raise ValueError("Liczba dni nie moze byc ujemna")
        return days
    except ValueError:
        return None


# Zadanie 1.4 – walidacja zamowien

def validate_orders(orders_list):
    """Sprawdza, czy kazde zamowienie ma order_id i amount, wypisuje podsumowanie."""
    invalid_count = 0
    for order in orders_list:
        missing_keys = []
        for key in ("order_id", "amount"):
            if key not in order:
                missing_keys.append(key)
        if missing_keys:
            invalid_count += 1
            print(f"Niepoprawne zamowienie: {order}. Brakujace pola: {missing_keys}")
    print(f"Liczba niepoprawnych zamowien: {invalid_count}")
    return invalid_count


# Zadanie 1.5 – wlasny wyjatek OrderNotFoundError

class OrderNotFoundError(Exception):
    """Wyjatek rzucany, gdy nie znaleziono zamowienia o danym id."""


def find_order(orders_list, order_id):
    """Zwraca slownik z zamowieniem lub rzuca OrderNotFoundError."""
    for order in orders_list:
        if order.get("order_id") == order_id:
            return order
    raise OrderNotFoundError(f"Nie znaleziono zamowienia o id {order_id}")


# Zadanie 1.6 – try / except / else / finally z plikiem

def read_file(path):
    """Probuje wczytac plik tekstowy i wypisuje podstawowe informacje."""
    f = None
    try:
        from pathlib import Path
        path_obj = Path(path)
        f = path_obj.open(encoding="utf-8")
        content = f.read()
    except FileNotFoundError:
        print(f"Plik {path} nie istnieje.")
    else:
        lines = content.splitlines()
        print(f"Plik {path} zostal poprawnie wczytany.")
        print(f"Liczba linii: {len(lines)}")
    finally:
        if f is not None:
            f.close()
````

````python
# Sekcja 2 – klasy i OOP

# Zadanie 2.1 – klasa Customer

class Customer:
    """Reprezentuje klienta w systemie zamowien."""

    def __init__(self, customer_id, name, country, segment, secret_note=None):
        self.customer_id = customer_id
        self.name = name
        self.country = country
        # umownie prywatny atrybut segmentu
        self._segment = segment
        # mocniej prywatna notatka
        self.__secret_note = secret_note

    def describe(self):
        """Wypisuje podstawowe informacje o kliencie."""
        print(
            f"Customer(id={self.customer_id}, name={self.name}, "
            f"country={self.country}, segment={self._segment})"
        )

    def get_segment(self):
        """Zwraca segment klienta."""
        return self._segment

    def _get_secret_note(self):
        """Przykladowy dostep do prywatnej notatki (metoda wewnetrzna)."""
        return self.__secret_note


# Zadanie 2.2 i 2.3 – klasa Order z licznikiem

class Order:
    """Reprezentuje pojedyncze zamowienie."""

    created_count = 0  # atrybut klasy – licznik utworzonych zamowien

    def __init__(self, order_id, customer_id, status, amount):
        self.order_id = order_id
        self.customer_id = customer_id
        self.status = status
        self.amount = amount
        Order.created_count += 1

    def is_paid(self):
        """Zwraca True jesli zamowienie jest oplacone."""
        return self.status == "paid"

    def apply_discount(self, percent):
        """Zastosuj procentowy rabat do kwoty zamowienia."""
        self.amount = self.amount * (1 - percent / 100)

    def __str__(self):
        return (
            f"Order(id={self.order_id}, customer_id={self.customer_id}, "
            f"status={self.status}, amount={self.amount})"
        )


# Zadanie 2.4 – dziedziczenie: OnlineOrder

class OnlineOrder(Order):
    """Zamowienie online z kosztem dostawy."""

    def __init__(self, order_id, customer_id, status, amount, shipping_cost):
        super().__init__(order_id, customer_id, status, amount)
        self.shipping_cost = shipping_cost

    def total_amount(self):
        """Zwraca laczna kwote zamowienia z dostawa."""
        return self.amount + self.shipping_cost


# Zadanie 2.5 – polimorfizm

class StandardOrder:
    def __init__(self, amount):
        self.amount = amount

    def calculate_total(self):
        return self.amount


class PromoOrder:
    def __init__(self, amount, discount_percent):
        self.amount = amount
        self.discount_percent = discount_percent

    def calculate_total(self):
        return self.amount * (1 - self.discount_percent / 100)


# Przykladowe uzycie polimorfizmu
example_orders = [StandardOrder(100), PromoOrder(200, 10)]
totals = [o.calculate_total() for o in example_orders]
totals
````

````python
# Sekcja 3 – metody magiczne i SOLID

# Zadanie 3.1 – __str__ (dla Customer juz dodane, dla Order dodane wyzej)
# Przyklad testu:
c_example = Customer(101, "Acme Corp", "PL", "B2B")
o_example = Order(1, 101, "paid", 100.0)
print(c_example)
print(o_example)


# Zadanie 3.2 i 3.3 – OrdersCollection

class OrdersCollection:
    """Kolekcja zamowien z zachowaniem podobnym do listy."""

    def __init__(self, orders_list):
        self._orders = orders_list

    def __len__(self):
        """Zwraca liczbe zamowien w kolekcji."""
        return len(self._orders)

    def __contains__(self, order_id):
        """Sprawdza, czy kolekcja zawiera zamowienie o danym ID."""
        return any(order.get("order_id") == order_id for order in self._orders)

    def __getitem__(self, index):
        """Zwraca zamowienie o podanym indeksie."""
        return self._orders[index]

    def __iter__(self):
        """Zwraca iterator po zamowieniach."""
        return iter(self._orders)


# Przykladowe testy OrdersCollection
collection_example = OrdersCollection(orders)
print("Liczba zamowien w kolekcji:", len(collection_example))
print("Czy jest zamowienie o ID=1?", 1 in collection_example)
print("Pierwsze zamowienie:", collection_example[0])
print("Kwoty zamowien z kolekcji:", [o["amount"] for o in collection_example])


# Zadanie 3.4 – Single Responsibility Principle

class OrdersGodObject:
    """Przykladowa klasa robiaca wszystko naraz – do analizy i refaktoryzacji."""

    def __init__(self, csv_path):
        self.csv_path = csv_path
        self.orders = []

    def load_validate_and_analyse(self):
        '''Wczytuje plik, waliduje dane i liczy statystyki.

        Ta metoda celowo lamie zasade Single Responsibility:
        1. Odpowiada za wczytanie pliku CSV.
        2. Odpowiada za walidacje danych.
        3. Odpowiada za liczenie i wypisywanie statystyk.

        W praktyce nalezaloby to podzielic na osobne klasy/funkcje.
        '''
        import csv
        from pathlib import Path

        path = Path(self.csv_path)
        if not path.exists():
            print(f"Plik {path} nie istnieje")
            return

        with path.open(encoding="utf-8") as f:
            reader = csv.DictReader(f)
            self.orders = list(reader)

        invalid = 0
        total = 0.0
        for order in self.orders:
            if "amount" not in order or order["amount"] == "":
                invalid += 1
                continue
            try:
                total += float(order["amount"])
            except ValueError:
                invalid += 1

        print(f"Wczytano zamowien: {len(self.orders)}")
        print(f"Niepoprawne rekordy: {invalid}")
        print(f"Suma kwot (bez blednych): {total}")


# Zadanie 3.5 – OrdersStats

class OrdersStats:
    """Liczy statystyki dla listy zamowien."""

    def __init__(self, orders_list):
        self.orders = orders_list

    def total_amount(self):
        """Zwraca sume kwot amount."""
        return sum(order.get("amount", 0) for order in self.orders)

    def count_paid(self):
        """Zwraca liczbe zamowien o statusie 'paid'."""
        return sum(1 for order in self.orders if order.get("status") == "paid")


# Szybki test OrdersStats na bazowych danych
stats_example = OrdersStats(orders)
print("Suma kwot:", stats_example.total_amount())
print("Liczba zamowien paid:", stats_example.count_paid())
````

````python
# Sekcja 4 – moduly, pakiety, pathlib, CSV

# Zadanie 4.1 i 4.2 – modul analytics_utils (do utworzenia jako osobny plik)
# Przykladowa zawartosc analytics_utils.py:
# def total_amount(orders):
#     return sum(order['amount'] for order in orders)
#
# Nastepnie w tym notatniku:
# from analytics_utils import total_amount as ta
# ta(orders)

# Zadanie 4.3 – mini pakiet ecommerce (tworzony poza notatnikiem)
# Struktura:
# ecommerce/
#   __init__.py (opcjonalnie)
#   orders.py
#   customers.py
#
# Przykladowy import:
# from ecommerce.orders import Order as PkgOrder
# o = PkgOrder(1, 101, 'paid', 100.0)
# print(o)

from pathlib import Path


        def list_csv_files(data_dir):
    """Zwraca liste sciezek do plikow CSV w katalogu data_dir."""
    data_dir = Path(data_dir)
    files = []
    for path in data_dir.glob('*.csv'):
        files.append(path)
    return files


def read_orders_csv(path):
    """Wczytuje plik CSV z zamowieniami do listy slownikow."""
    from pathlib import Path as _Path

    path = _Path(path)
    orders_from_file = []
    with path.open(encoding='utf-8') as f:
        header_line = f.readline().strip()
        if not header_line:
            return orders_from_file
        header = header_line.split(',')
        for line in f:
            values = line.strip().split(',')
            if not values or values == ['']:
                continue
            row = dict(zip(header, values))
            # konwersje typow
            if 'order_id' in row and row['order_id'] != '':
                row['order_id'] = int(row['order_id'])
            if 'customer_id' in row and row['customer_id'] != '':
                row['customer_id'] = int(row['customer_id'])
            if 'amount' in row and row['amount'] != '':
                row['amount'] = float(row['amount'])
            orders_from_file.append(row)
    return orders_from_file


class OrdersLoader:
    """Wczytuje zamowienia z wielu plikow CSV w katalogu data_dir."""

    def __init__(self, data_dir):
        self.data_dir = Path(data_dir)

    def load_all(self):
        """Zwraca liste wszystkich zamowien z plikow CSV (orders_*.csv)."""
        all_orders = []
        for path in self.data_dir.glob('orders_*.csv'):
            orders_from_file = read_orders_csv(path)
            all_orders.extend(orders_from_file)
        return all_orders


def main():
    """Mini projekt: wczytaj zamowienia, zbuduj kolekcje i policz statystyki."""
    loader = OrdersLoader('data')
    orders_all = loader.load_all()
    print(f"Wczytano lacznie zamowien: {len(orders_all)}")
    collection = OrdersCollection(orders_all)
    stats = OrdersStats(orders_all)
    print(f"Liczba zamowien w kolekcji: {len(collection)}")
    print(f"Suma kwot: {stats.total_amount()}")
    print(f"Liczba zamowien oplaconych: {stats.count_paid()}")


if __name__ == '__main__':
    # Uruchomienie mini-projektu – odkomentuj po przygotowaniu plikow CSV w katalogu data
    # main()
    pass
````
