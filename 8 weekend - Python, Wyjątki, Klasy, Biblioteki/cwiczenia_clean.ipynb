### Zadanie 1.6 – try / except / else / finally przy pracy z plikiem
- Poziom: 2 / 5

- Biznes: wczytujesz plik CSV z zamowieniami i chcesz miec czytelny komunikat, gdy plik nie istnieje, oraz pewnosc, ze plik jest zamkniety.
- Technicznie: napisz funkcje `read_file(path)`, ktora uzywa `try / except / else / finally`.

````python
# Zadanie 1.6 – try / except / else / finally z plikiem

def read_file(path):
    '''Probuje wczytac plik tekstowy i wypisuje podstawowe informacje.'''
    f = path
    try:
        with open(f, 'r') as file:
            content = file.read()
    except FileNotFoundError:
        print(f'Plik {f} nie zostal znaleziony.')
    else:
        print(f'Liczba znakow w pliku: {len(content)}')
    finally:
        print('EOF')

read_file("data/customers.csv")
read_file("data2/customers.csv")
````

### Zadanie 2.5 – Polimorfizm: roznorodne zamowienia
- Poziom: 3 / 5
- Czas: ok. 20 minut
- Biznes: masz rozne typy zamowien, ale chcesz liczyc ich wartosc wspolnym interfejsem.
- Technicznie: zdefiniuj dwie klasy z metoda `calculate_total` i przetestuj iteracje po liscie obiektow.

````python
# Zadanie 2.5 – polimorfizm

class StandardOrder:
    def __init__(self, amount):
        self.amount = amount

    def calculate_total(self):
        return self.amount

class PromoOrder:
    def __init__(self, amount, discount_percent):
        self.amount = amount
        self.discount_percent = discount_percent

    def calculate_total(self):
        return self.amount * self.discount_percent


lista_zamowien = [
    StandardOrder(amount=100.0),
    PromoOrder(amount=200.0, discount_percent=0.8),
    StandardOrder(amount=50.0),
    PromoOrder(amount=150.0, discount_percent=0.9)
]

laczna_wartosc = 0.0

print("--- Przetwarzanie zamówień ---")
for zamowienie in lista_zamowien:
    # Polimorfizm w akcji: Python sam wie, którą metodę wywołać
    wartosc_pozycji = zamowienie.calculate_total()
    laczna_wartosc += wartosc_pozycji
    
    # Wyświetlamy typ obiektu oraz jego obliczoną wartość
    typ_klasy = zamowienie.__class__.__name__
    print(f"Typ: {typ_klasy:<15} | Wartość: {wartosc_pozycji:>6} zł")

print("-" * 38)
print(f"Łączna wartość wszystkich zamówień: {laczna_wartosc} zł")
````
