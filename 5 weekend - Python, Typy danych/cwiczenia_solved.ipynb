## Ćwiczenie 4: PEP8 – czytelność wyrażeń

**Opis biznesowy:**

**Poziom trudności:** 2/5

```python
# Rozwiązanie ćwiczenia 4

# Niechlujny kod:
# totalPrice=100+20*0.23
# CustomerID=1001
# FinalRESULT=totalPrice/2

# Poprawiona wersja zgodnie z PEP8
total_price_net = 100.0
extra_fee = 20.0
fee_rate = 0.23
customer_id = 1001

total_price_with_fee = total_price_net + extra_fee * fee_rate
final_result = total_price_with_fee / 2

print("ID klienta:", customer_id)
print("Wynik końcowy:", final_result)
```

## Ćwiczenie 20: Operacje na zbiorach – klienci z różnych kampanii

**Opis biznesowy:**
- Masz dwie kampanie marketingowe A i B.
- Chcesz wiedzieć:
  - ilu klientów wzięło udział w kampanii A,
  - ilu tylko w B, lub w obu jednocześnie.

**Cel techniczny:**
- Użycie operacji: suma (`|`), przecięcie (`&`), różnica (`-`).

**Poziom trudności:** 3/5  
**Szacowany czas:** 10 min


```python
# Rozwiązanie ćwiczenia 20

campaign_a_customers = {1001, 1002, 1003, 1005}
campaign_b_customers = {1003, 1004, 1005, 1006}

both_campaigns = campaign_a_customers & campaign_b_customers
only_a = campaign_a_customers - campaign_b_customers
only_b = campaign_b_customers - campaign_a_customers
any_campaign = campaign_a_customers | campaign_b_customers

print("Klienci w obu kampaniach:", both_campaigns, "(liczba:", len(both_campaigns), ")")
print("Tylko kampania A:", only_a, "(liczba:", len(only_a), ")")
print("Tylko kampania B:", only_b, "(liczba:", len(only_b), ")")
print("W co najmniej jednej kampanii:", any_campaign, "(liczba:", len(any_campaign), ")")
```

## Ćwiczenie 24: Aktualizacja słownika – nowa transakcja klienta

**Opis biznesowy:**
- Klient robi kolejne zakupy.
- Chcesz zaktualizować jego łączną kwotę zakupów.

**Cel techniczny:**
- Aktualizacja wartości w `dict` oraz użycie `update()`.

**Poziom trudności:** 3/5  
**Szacowany czas:** 10 min


```python
# Rozwiązanie ćwiczenia 24

customer_spent = {
    1001: 4300.0,
    1002: 80.0,
    1003: 150.0,
}

new_transaction_customer = 1001
new_transaction_value = 500.0

# Aktualizacja wartości w słowniku: dodajemy nową transakcję do dotychczasowej sumy
previous_total = customer_spent.get(new_transaction_customer, 0.0)
customer_spent[new_transaction_customer] = previous_total + new_transaction_value

print("Po aktualizacji klienta 1001:", customer_spent)

# Użycie update() do dodania nowego klienta
new_entries = {1004: 300.0}
customer_spent.update(new_entries)
print("Po dodaniu nowego klienta 1004:", customer_spent)
```

```python
# Rozwiązanie ćwiczenia 26 (mini-projekt)

# 1. Lista produktów i cen
products = ["Laptop", "Mouse", "Keyboard"]
prices = [3500.0, 80.0, 150.0]

# 2. Zbiór kategorii
categories = {"electronics", "accessories"}

# 3. Słownik: client_id -> total_spent
customer_spent = {
    1001: 3500.0,  # kupił laptopa
    1002: 80.0,    # kupił myszkę
}

print("Początkowe produkty:", products)
print("Początkowe ceny:", prices)
print("Początkowe kategorie:", categories)
print("Początkowi klienci:", customer_spent)

# 4. Dodaj nowy produkt i zaktualizuj odpowiednie struktury
new_product = "Monitor"
new_product_price = 900.0
products.append(new_product)
prices.append(new_product_price)
categories.add("electronics")  # już istnieje, ale to nie problem

print("\nPo dodaniu produktu:")
print("Produkty:", products)
print("Ceny:", prices)
print("Kategorie:", categories)

# 5. Dodaj nowego klienta z pierwszą transakcją
new_customer_id = 1003
new_customer_total = new_product_price + prices[1]  # kupił monitor + myszkę
customer_spent[new_customer_id] = new_customer_total

print("\nPo dodaniu nowego klienta:")
print("Klienci i wydatki:", customer_spent)

# 6. Oblicz prostą statystykę (średnia wartość koszyka)
total_spent_sum = (
    customer_spent[1001]
    + customer_spent[1002]
    + customer_spent[1003]
)
number_of_customers = 3
average_cart_value = total_spent_sum / number_of_customers

# Krótkie podsumowanie
summary = (
    "Podsumowanie mini-modelu sklepu:\n"
    f"\tLiczba produktów: {len(products)}\n"
    f"\tLiczba kategorii: {len(categories)}\n"
    f"\tLiczba klientów: {number_of_customers}\n"
    f"\tŚrednia wartość koszyka: {average_cart_value} PLN\n"
)

print("\n" + summary)
```
